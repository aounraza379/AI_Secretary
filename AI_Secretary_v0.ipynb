```python
import re
from PIL import Image, ImageOps
import pytesseract

# 1. Run the best OCR pipeline setup
original_image = Image.open("/content/huawei_course_poster.jpg")
resized_image = original_image.resize((original_image.width * 2, original_image.height * 2))
grayscale_image = ImageOps.grayscale(resized_image)

# 2. Extract and store the text
ocr_text = pytesseract.image_to_string(grayscale_image)
print("--- RAW OCR TEXT EXTRACTED ---")
print(ocr_text)
print("------------------------------\n")

# 3. Minimal Next Experiment: Extract key information using simple Python rules
print("--- AI SECRETARY: EXTRACTED KEY INFO ---")

# Extract Phone Numbers using regex
phone_pattern = r"\b\d{4}[-\s]?\d{7}\b"
phones = re.findall(phone_pattern, ocr_text)
print(f"Detected Contact Numbers: {list(set(phones)) if phones else 'None Found'}")

# Extract lines containing common keywords
keywords = ["duration", "eligibility", "fee", "free"]
for line in ocr_text.split('\n'):
    clean_line = line.strip()
    if any(kw in clean_line.lower() for kw in keywords):
        print(f"Relevant Line Found: {clean_line}")
```

```python
import json
from google import genai
from google.colab import userdata

# 1. Initialize stable client using the Colab Secret Key securely
api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)

# 2. Design the prompt to structure our raw OCR text
prompt = f"""
Analyze the following messy OCR text from an informational poster and extract the key details in a clean JSON format.
Include fields like: program_name, eligibility, duration, certification_fee, contact_numbers, and other_important_details.

OCR Text:
{ocr_text}
"""

# 3. Call the recommended gemini-3.8-flash model (Retrying transient 503)
response = client.models.generate_content(
    model='gemini-3.8-flash',
    contents=prompt,
    config={
        'response_mime_type': 'application/json'
    }
)

# 4. Display the structured results
structured_data = json.loads(response.text)
print(json.dumps(structured_data, indent=2))
```

```python
def format_secretary_summary(data):
    """
    Converts structured JSON data into a clean, human-readable summary
    suitable for instant display in chat interfaces, emails, or markdown dashboards.
    """
    summary = []
    summary.append(f"📋 **PROGRAM OUTLINE**: {data.get('program_name', 'N/A')}")
    summary.append("─" * 50)

    eligibility = data.get('eligibility', {})
    edu = eligibility.get('education', 'N/A')
    age = eligibility.get('age_limit', 'N/A')
    summary.append(f"🎓 **Eligibility**:\n  • **Education**: {edu}\n  • **Age**: {age}")

    summary.append(f"⏳ **Duration**: {data.get('duration', 'N/A')}")
    summary.append(f"💰 **Cost**: {data.get('certification_fee', 'N/A')}")

    contacts = ", ".join(data.get('contact_numbers', []))
    summary.append(f"📞 **Contact Info**: {contacts if contacts else 'N/A'}")

    details = data.get('other_important_details', {})
    summary.append(f"💡 **Highlights**:\n  • **Initiative**: {details.get('initiative', 'N/A')}\n  • **Certificate**: {details.get('certification_type', 'N/A')}\n  • **Recognition**: {details.get('recognition', 'N/A')}")

    return "\n\n".join(summary)

# Display the beautifully formatted human-readable summary
formatted_text = format_secretary_summary(structured_data)
print("--- AI SECRETARY HUMAN READABLE OUTPUT ---\n")
print(formatted_text)
```

```python
from datetime import datetime, timedelta

def generate_ics_file(data, filename="event.ics"):
    """
    Converts our parsed JSON data into a standard .ics file structure.
    This file can be opened directly in Google Calendar, Outlook, or Apple Calendar.
    """
    # Since the poster mentions '3 Months' duration, we'll schedule a placeholder start/end date.
    # In a fully realized app, you would parse the start dates dynamically if they are on the poster.
    start_date = datetime.now() + timedelta(days=7) # Assume starts in a week
    end_date = start_date + timedelta(days=90)      # 3 Months duration

    start_str = start_date.strftime("%Y%m%dT090000")
    end_str = end_date.strftime("%Y%m%dT170000")
    timestamp = datetime.now().strftime("%Y%m%dT%H%M%SZ")

    # Compose the ICS content
    ics_content = f"""BEGIN:VCALENDAR
VERSION:2.0
PRODID:-//AI Secretary//NONSGML Event//EN
BEGIN:VEVENT
UID:{timestamp}-secretary@example.com
DTSTAMP:{timestamp}
DTSTART;TZID=Asia/Karachi:{start_str}
DTEND;TZID=Asia/Karachi:{end_str}
SUMMARY:{data.get('program_name', 'Education Course')}
DESCRIPTION:Generated by your AI Secretary.\n\nEligibility: {data.get('eligibility', {}).get('education', 'N/A')}\nFee: {data.get('certification_fee', 'N/A')}\nContact: {", ".join(data.get('contact_numbers', []))}\n\nInitiative: {data.get('other_important_details', {}).get('initiative', 'N/A')}
LOCATION:NAVTTC Multan / Tech Heaven
END:VEVENT
END:VCALENDAR"""

    # Write to a file in Google Colab
    with open(filename, "w", encoding="utf-8") as f:
        f.write(ics_content.strip())

    print(f"[Success] Generated calendar file: {filename}")
    print("You can download this file from the left-hand files sidebar in Colab and open it directly in your calendar application!")

# Run the generator using our extracted JSON data
generate_ics_file(structured_data, "Huawei_Tech_Essentials.ics")
```

```python
import urllib.parse
from datetime import datetime, timedelta

def generate_google_calendar_link(data):
    """
    Generates a direct, shareable Google Calendar 'templates' link.
    When clicked, this immediately populates the user's Google Calendar with all details.
    """
    base_url = "https://calendar.google.com/calendar/render?action=TEMPLATE"

    # Title
    text = data.get('program_name', 'Education Course')

    # Set up placeholder dates (starts in 7 days, lasting 3 months)
    start_date = datetime.now() + timedelta(days=7)
    end_date = start_date + timedelta(days=90)
    dates_str = f"{start_date.strftime('%Y%m%dT090000')}/{end_date.strftime('%Y%m%dT170000')}"

    # Combine key elements into the description parameter
    eligibility = data.get('eligibility', {})
    details = data.get('other_important_details', {})

    description_lines = [
        f"🎓 Eligibility: {eligibility.get('education', 'N/A')} (Age: {eligibility.get('age_limit', 'N/A')})",
        f"💰 Fee: {data.get('certification_fee', 'N/A')}",
        f"📞 Contacts: {', '.join(data.get('contact_numbers', []))}",
        f"💡 Certificate: {details.get('certification_type', 'N/A')}",
        f"⭐ Initiative: {details.get('initiative', 'N/A')}",
        "\nGenerated automatically by your AI Secretary!"
    ]
    description = "\n".join(description_lines)

    location = "NAVTTC Multan / Tech Heaven"

    # URL encode all parameters safely
    params = {
        "text": text,
        "dates": dates_str,
        "details": description,
        "location": location,
        "sf": "true",
        "output": "xml"
    }

    encoded_params = urllib.parse.urlencode(params)
    return f"{base_url}&{encoded_params}"

# Generate our link!
gcal_link = generate_google_calendar_link(structured_data)
print("--- PRODUCTION GOOGLE CALENDAR LINK ---\n")
print(gcal_link)
print("\n----------------------------------------")
print("Click this link to test adding it directly to your personal calendar!")
```
